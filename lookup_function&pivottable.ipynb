"Below Target" otherwise
Also calculate the percentage of records where the target is achieved.

 - Formula Used (Performance Status):
=IF(I2>=L2,"Target Achieved","Below Target")
Formula Used (% Target Achieved):
=COUNTIF(O2:O1001,"Target Achieved")/COUNTA(O2:O1001)
Explanation:
The IF function compares PaymentAmount with TargetRevenue for each record.
If PaymentAmount is greater than or equal to TargetRevenue, the record is marked as "Target Achieved"; otherwise, it is marked as "Below Target".
The COUNTIF formula counts the number of successful records and divides it by the total number of records to calculate the achievement percentage.
